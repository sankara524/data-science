# Exploratory Data Analysis

## 1. Univariate Analysis

### Which Batting Team Played Most Balls

```python
df['batting_team'].value_counts()
```

## Bar Plot

```python
plt.figure(figsize = (15,10))
df['batting_team'].value_counts().plot(kind = 'bar')
plt.title('Balls Played by Each Team from 2008 to 2025', c='red')
plt.xlabel('Batting Teams', c='red')
plt.ylabel('No.of Balls', c='red')
plt.grid(True)
plt.xticks(rotation = 90)
plt.show()
```

## Business Insights:

- The Analysis shows that the Mumbai Indians team face the highest number of balls, This suggest that the team has participated in a large number of innings or matches over the years.

### Recommendations:

- Analyze this team's batting strategy to identify successful techniques.
- Other teams can benchmark their batting approach against this team's consistency.
- Coaches can study batting partnerships and innings-building patterns.

### Distribution of Total Runs

```python
df['runs_total'].value_counts()
```

## Histogram

```python
plt.figure(figsize = (8, 6))
sns.histplot(df['runs_total'], bins = 8)
plt.title('Distribution of Runs per Ball', c='red')
plt.xlabel('Total Runs Per Ball', c='red')
plt.ylabel('Total No.of Balls', c='red')
plt.grid(True)
plt.show()
```

## Business Insights:

- The distribution of total runs per ball indicates that 0 and 1 run outcomes occur most frequently, while boundaries (4s and 6s) and high-scoring deliveries are comparatively rare.

### Recommendation:

- Batting coaches should focus on increasing strike rotation to reduce dot balls.
- Bowling teams should continue strategies that restrict singles and boundaries.
- Teams should identify overs where scoring opportunities are highest.

### Distribution of Runs with Box Plot

```python
plt.figure(figsize = (8, 6))
sns.boxplot(x = df['runs_total'])
plt.xlabel('Total Runs Per Ball', c='red')
plt.show()
```

```python
plt.figure(figsize = (8, 6))
sns.boxplot(y = df['runs_total'])
plt.ylabel('Total Runs Per Ball', c='red')
plt.show()
```

### Business Insights:

- The Box Plot reveals that the majority of deliveries result in low run values, while a small number of deliveries produce significantly higher scores.

### Recommendation:
- Bowling units should analyze the situations that lead to high-scoring deliveries.
- Batting teams should identify bowlers and match situations where boundaries are more likely.
- Teams can improve death-over planning to control or maximize scoring.

## 2. Bivariate Analysis

### Total Runs by Batting Team

```python
team_runs = df.groupby('batting_team')['runs_total'].sum().sort_values(ascending = False)
team_runs
```

## Bar Plot

```python
team_runs.sort_values(ascending = False).plot(
    kind = 'bar',
    figsize = (8,6))
plt.title('Total Runs by Batting Team', c='red')
plt.xlabel('Batting Team', c='red')
plt.ylabel('Total Runs', c='red')
plt.show()
```

### Business Insights:

- Mumbai Indians has scored the highest total runs among all batting teams in the dataset. This indicating consistent batting performance across multiple IPL seasons.

### Recommendation:

- Lower-performing teams should review the batting strategies of top-performing teams.
- Coaches should emphasize partnerships, strike rotation, and boundary conversion.
- Team management can use these comparisons to identify areas for improvement.

### Total Runs by Season

```python
season_runs = df.groupby('season')['runs_total'].sum()#.sort_values(ascending = False)
season_runs
```

## Line Plot

```python
season_runs.plot(
    kind = 'line',
    marker = 'o',
    figsize = (8,6))
plt.title('Season Wise Total Runs', c='red')
plt.xlabel('Season', c='red')
plt.ylabel('Total Runs', c='red')
plt.show()
```

## Business Insights:

- Total runs have increased over the years, suggesting a trend toward more aggressive batting and higher-scoring matches.

### Recommendation:

- Teams should prepare bowling strategies for increasingly high-scoring matches.
- Batting units should continue adapting to modern aggressive playing styles.
- Match analysts should monitor scoring trends when planning team strategies.

## 3. Multi Variate Analysis

### Comparing how teams performed across seasons.

```python
pivot = df.pivot_table(
    index = 'season',
    columns = 'batting_team',
    values = 'runs_total',
    aggfunc = 'sum'
)
pivot
```

## Histogram

```python
plt.figure(figsize = (8,6))
sns.heatmap(pivot)
plt.title('Seasons vs Batting Team Runs', c='red')
plt.xlabel('Batting Team', c = 'red')
plt.ylabel('Seasons', c='red')
plt.show()
```

### Business Insights:

- The heatmap highlights the variation in team performance across different IPL seasons. Some teams maintained consistently high scoring across multiple seasons, while others showed fluctuations in performance.

### Recommendation:

- Teams with inconsistent performance should identify seasons with successful results and replicate those strategies.
- Management should evaluate long-term player performance and squad planning.
- Coaches should use seasonal performance analysis to improve future planning.
