### 3. Write a program where class Account has a method deposit(). Inherit it in class SavingsAccount and add a method add_interest(). Demonstrate both methods using a SavingsAccount object.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposit Amount: {amount}")
        print(f"Total Balance: {self.balance}")

class SavingsAccount(Account):
    def add_interest(self, rate):
        interest = self.balance * rate / 100
        self.balance += interest
        print(f"Interest: {interest}")
        print(f"After Deposit: {self.balance}")

account1 = SavingsAccount(100000)
account1.deposit(10000)
account1.add_interest(10)
```

### 10. Create a base class Shape with a method area(). Inherit two classes Square and Circle from it. Implement area calculation in both and demonstrate with objects.

```python
class Shape:
    def area(self):
        print("Area Calculation")

class Square(Shape):
    def __init__(self, square):
        self.square = square
    def area(self):
        print(f"Area of Square: {self.square * self.square}")   # square = x*x

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        print(f"Area of Circle: {3.14 * self.radius * self.radius}")  # Radius of circle = 3*radius*radius

square1 = Square(5)
circle1 = Circle(5)

square1.area()
circle1.area()
```

### 16. Write a Python program that demonstrates polymorphism using a function area() that works for both circle and rectangle.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print(f"The area of the Circle is: {3.14 * self.radius * self.radius}")

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print(f"The area of the Rectangle is: {self.length * self.width}")

circle1 = Circle(5)
rectangle1 = Rectangle(10, 20)

circle1.area()
rectangle1.area()
```

## **Part C: Encapsulation**

### 21. Create a class BankAccount with private variables _balance. Provide methods to deposit and withdraw money, and show how encapsulation controls access.

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        self._balance = self._balance + amount
        print(f"Amount is Deposited: {amount}")

    def withdraw(self, amount):
        if amount <= self._balance:
            self._balance = self._balance - amount
            print(f"Amount Withdraw: {amount}")
        else:
            print("Insufficient Balance")

    def show_balance(self):
        print(f"Current Balance Amount: {self._balance}")

account1 = BankAccount(1000)

account1.show_balance()

account1.deposit(1000)
account1.show_balance()

account1.withdraw(500)
account1.show_balance()
```

### 22. Create a class Laptop with private attributes _brand and _price. Use encapsulation to print details only if the user has proper access.

```python
class Laptop:
    def __init__(self, brand, price):
        self._brand = brand
        self._price = price

    def show_details(self, password):
        if password == 12345:
            print(f"Brand: {self._brand}")
            print(f"Price: {self._price}")
        else:
            print("This Password is Incorrect, Please Try Again..")

laptop1 = Laptop("HP", 40000)

laptop1.show_details(12345)
```
