### 1. Strong Number

A Strong Number is a number in which the sum of the factorials of its digits is equal to the number itself.

Example:
 145 → 1! + 4! + 5! = 1 + 24 + 120 = 145

Task:
 Write a Python program to check whether a number is a Strong Number.

Input:
 145

Output:
 145 is a Strong Number


```python
num = 145
temp = num
total = 0
while temp > 0:
    digit = temp%10
    fact = 1
    for i in range(1, digit+1):
        fact *= i
    total += fact
    temp = temp // 10
if total == num:
    print("Number is Strong", num)
else:
    print("Number is not Strong", num)
```

### 2. Prime Number

A prime number is a number greater than 1 that has no divisors other than 1 and itself.

Example:
 7 → Divisors are 1 and 7 only.

Task:
 Write a Python program to check whether a number is a Prime Number.

Input:
 7

Output:
 7 is a Prime Number

```python
num = 7
if num <= 1:
    print(f"{num} is not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print(f"{num} is not a Prime Number")
            break
    else:
            print(f"{num} is a Prime Number")
```

### 6. Armstrong Number

An Armstrong number is a number in which the sum of its digits raised to the power of the number of digits equals the number itself.

Example:
 153 → 1³ + 5³ + 3³ = 153

Task:
 Write a Python program to check whether a number is an Armstrong Number.

Input:
 153

Output:
 153 is an Armstrong Number

```python
num = 153
temp = num
result = 0

while temp > 0:
    digit = temp % 10
    result += digit ** 3
    temp = temp // 10
if result == num:
    print(f"{result}  is an Armstrong Number")
else:
    print(f"{result}  is not an Armstrong Number")
```

### 8. Sum of Digits

The sum of digits of a number is obtained by adding all its digits together.

### 9. Perfect Number

A perfect number is a number that is equal to the sum of its proper divisors (excluding itself).

Example:
 28 → 1 + 2 + 4 + 7 + 14 = 28

Task:
 Write a Python program to check whether a number is a Perfect Number.

Input:
 28

Output:
 28 is a Perfect Number

```python
num = 28
sum1 = 0
for i in range(1, num):
    if num % i == 0:
        sum1 += i
if sum1 == num:
    print(f"{num} is a Perfect Number")
else:
    print(f"{num} is not a Perfect Number")
```

### 10. GCD (Greatest Common Divisor)

The GCD of two numbers is the largest number that divides both of them exactly.

Example:
 GCD of 20 and 28 is 4.

Task:
 Write a Python program to find the GCD of two numbers.

Input:
 20 28

Output:
 GCD = 4

```python
x = 20
y = 28
for i in range(1, min(x,y)+1):
    if (x % i == 0) and (y % i == 0):
        GCD = i
print(f"GCD = {GCD}")
```

### 11. LCM (Least Common Multiple)

The LCM of two numbers is the smallest number divisible by both.

Example:
 LCM of 4 and 6 is 12.

Task:
 Write a Python program to find the LCM of two numbers.

Input:
 4 6

Output:
 LCM = 12

```python
x, y = 4, 6
n = 1
while True:
    if (n % x == 0) and (n % y == 0):
        print(f"LCM = {n}")
        break
    n += 1
```

### 13. Harshad Number

A Harshad number is divisible by the sum of its digits.

Example:
 18 → Sum of digits = 1+8 = 9 → 18 ÷ 9 = 2 (divisible).

Task:
 Write a Python program to check whether a number is a Harshad Number.

Input:
 18

Output:
 18 is a Harshad Number

```python
num = 18
str1 = str(num)
lst = list(str1)
result = 0
for i in range(len(lst)):
    result += int(lst[i])
if num % result == 0:
    print(f"{num} is a Harshad Number")
else:
    print(f"{num} is not a Harshad Number")
```

### 16. HCF & LCM of More Than Two Numbers

HCF is the highest number dividing all numbers exactly, and LCM is the smallest number divisible by all.

Example:
 Numbers: 12, 18, 24 → HCF = 6, LCM = 72

Task:
 Write a Python program to find the HCF and LCM of more than two numbers.

Input:
 12 18 24

Output:
 HCF = 6, LCM = 72

```python
a,b,c = 12,18,24

# HCF
GCD = 1
for i in range(1, min(a,b,c)+1):
    if (a % i == 0) and (b % i == 0) and (c % i == 0):
        GCD = i

# LCM
num = max(a, b, c)

while True:
    if (num % a == 0) and (num % b == 0) and (num % c == 0):
        LCM = num
        break
    num += 1

print(f"HCF = {GCD}, LCM = {LCM}")
```

### 19. Leap Year

A leap year is divisible by 4 but not by 100, except if also divisible by 400.

Example:
 2024 → Divisible by 4, not by 100 → Leap Year.

Task:
 Write a Python program to check whether a year is a Leap Year.

Input:
 2024

Output:
 2024 is a Leap Year

```python
year = int(input("Enter Year: "))
if((year % 4 == 0) and (year % 100 != 0)) or (year % 400 == 0):
    print(f"{year}  is a Leap Year")
else:
    print(f"{year} is not a Leap Year")
```

### 20. Perfect Square

A number is a perfect square if its square root is an integer.

Example:
 49 → √49 = 7 (integer).

Task:
 Write a Python program to check whether a number is a Perfect Square.

Input:
 49

Output:
 49 is a Perfect Square

```python
import math

n = int(input("Enter number: "))

num = int(math.sqrt(num))
if num * num == n:
    print(f"{n} is a Perfect Square")
else:
    print(f"{n} is not a Perfect Square")
```
