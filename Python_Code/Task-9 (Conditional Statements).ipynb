### 1. A bookstore gives discounts based on purchase amount:

- If bill ≥ 5000 → "20% Discount"

- If bill ≥ 3000 → "10% Discount"

- If bill ≥ 1000 → "5% Discount"

- Otherwise → "No Discount"

- Input: bill = 3500

- Expected Output: 10% Discount


```python
bill = int(input("bill = "))
if (bill >= 5000):
    print("20% Discount")
elif (bill >= 3000):
    print("10% Dicount")
elif (bill >= 1000):
    print("5% Discount")
else:
    print("No Discount")
```

### 1. Email Format Validation

Write a Python program to check if an email is valid based on these rules:

- Must contain exactly one @ symbol.

- Must end with .com or .in.

- The username (before @) should contain only letters, numbers, dots, or underscores (no spaces or special symbols).

- The domain (after @) should contain only letters (e.g., gmail.com, yahoo.in).

Print "Valid Email" if all conditions are satisfied, else "Invalid Email".

```python
email = input("Enter Email = ")

count1 = email.count("@") == 1
count2 = email.split("@")
count3 = email.endswith(".com") or email.endswith(".in")
count4 = email.replace(".", "").replace("_", " ").isalnum()
count5 = email.split("@") 

if count1 and count2 and count3 and count4 and count5:
    print("Valid Email")
else:
    print("Invalid Email")
```

### 2. Strong Password Check

Take a password from the user and check:

- At least 8 characters

- Contains at least 1 uppercase letter

- Contains at least 1 digit

- Contains at least 1 special character (@, #, $, %)

If all conditions match → print "Strong Password" else "Weak Password".

```python
password = input("Enter a Password:")

length = len(password) >= 8
upper = password.upper() != password
digit = not password.isalpha()
special = not password.isalnum()

if length and upper and digit and special:
    print("Strong Password")
else:
    print("Weak Password")
```

### 3. Company Email Validation

Accept an email input and check:

- Must end with @company.com (only company employees can log in).

- Username before @ must start with a letter.

- Passwords should follow Q2 strong password rules.

If valid → "Access Granted", else "Access Denied".

```python
email = input("Enter Email:")
password = input("Enter Password:")

employee = email.endswith("@company.com")
user_name = email.split("@")[0]

length = len(password) >= 8
upper = password.upper() != password
digit = not password.isalpha()
special = not password.isalnum()

if employee and user_name and length and upper and digit and special:
    print("Access Granted")
else:
    print("Access Denied")
```
