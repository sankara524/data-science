## **Part A : Dictionary Comprehension**

### 1. Create a dictionary where keys are numbers from 0 to 4 and values are their squares.
 - Input: range(5)
 - Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

```python
dicts1 = {i: i*i for i in range(5)}
print(dicts1)
```

### 2. Create a dictionary from a list of fruits, where the fruit is the key and its length is the value.
 - Input: ["apple", "banana", "cherry"]
 - Output: {'apple': 5, 'banana': 6, 'cherry': 6}

```python
list1 = ["apple", "banana", "cherry"]
print(list1)

length = {i: len(i) for i in list1}
print(length)
```

### 3. Create a dictionary where keys are numbers from 1 to 3 and values are the string "hello".
 - Input: range(1, 4)
 - Output: {1: 'hello', 2: 'hello', 3: 'hello'}

```python
string1 = "hello"
print(string1)

num = {i: string1 for i in range(1, 4)}
print(num)
```

### 4. Create a dictionary from a list of numbers, where keys are the numbers and values are True if the number is even, False otherwise.
 - Input: [1, 2, 3, 4]
 - Output: {1: False, 2: True, 3: False, 4: True}

```python
list2 = [1,2,3,4]
print(list2)

even_odd = {i: 'True' if i%2 == 0 else 'False' for i in list2 }
print(even_odd)
```

### 5. Create a dictionary where keys are numbers from 0 to 2 and values are their cubes.
 - Input: range(3)
 - Output: {0: 0, 1: 1, 2: 8}

```python
num = {i: i**3 for i in range(0,3)}
print(num)
```

### 6. Create a dictionary from a list of names, where the name is the key and its first letter is the value.
 - Input: ["Alice", "Bob", "Charlie"]
 - Output: {'Alice': 'A', 'Bob': 'B', 'Charlie': 'C'}

```python
names = ["Alice", "Bob", "Charlie"]
print(names)

name1 = {i: i[0] for i in names}
print(name1)
```

### 7. Create a dictionary where keys are numbers from 1 to 5, and values are "Even" if the number is even, "Odd" if it's odd.
 - Input: range(1, 6)
 - Output: {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
numbers = {i: "Odd" if i%2 != 0 else "Even" for i in range(1, 6)}
print(numbers)
```

### 8. Create a dictionary from a list of words, where the word is the key and its uppercase version is the value.
 - Input: ["cat", "dog", "fish"]
 - Output: {'cat': 'CAT', 'dog': 'DOG', 'fish': 'FISH'}

```python
list3 = ["cat", "dog", "fish"]
print(list3)

case = {i: i.upper() for i in list3}
print(case)
```

### 9. Create a dictionary where keys are numbers from 10 to 12 and values are the number itself.
 - Input: range(10, 13)
 - Output: {10: 10, 11: 11, 12: 12}

```python
numbers = {i: i for i in range(10, 13)}
print(numbers)
```

### 10. Create a dictionary from a list of temperatures, where the temperature is the key and True if it's above 20, False otherwise.
 - Input: [15, 22, 18, 25]
 - Output: {15: False, 22: True, 18: False, 25: True}

```python
list4 = [15, 22, 18, 25]
print(list4)

num = {i: "False" if i < 20 else "True" for i in list4}
print(num)
```

## **Part B : break, pass, continue**

### 1. Print numbers from 0 to 4. Stop the loop when the number is 5.
 - Input: for i in range(10):
 - Output:
   
   0

   1

   2

   3

   4

```python
for i in range(0, 10):
    if (i == 5):
        break
        i= i+1
    print(i)
```

 - Input: for i in range(8):

### 6. Iterate through a list of names. If a name is "Bob", do nothing and continue.
 - Input: names = ["Alice", "Bob", "Charlie"]
 - Output:

Alice

Charlie

```python
names = ["Alice", "Bob", "Charlie"]
print(names)

for i in names:
    if (i == "Bob"):
        continue
    print(i)
```

### 8. Iterate through a string. If the character is 'a', skip it.
 - Input: word = "banana"
 - Output:

b

n

n

```python
word = "banana"
print(word)

for i in word:
    if (i == "a"):
        continue
    print(i)
```

### 10. Iterate through a list of numbers. If a number is greater than 10, stop the loop.
 - Input: numbers = [5, 8, 12, 3, 15]

5

```python
numbers = [5, 8, 12, 3, 15]
for i in numbers:
    if (i > 10):
        break
    print(i)
```
