## **Task-4 (List)**

### 1. Change the second element of a list to 100.

```python
lists = [1, 2, 3]
print(lists)

lists[1] = 100
print(lists)
```

### 2. Append an element to a list.

```python
lists = [1,2,3]
print(lists)

lists.append(4)
print(lists)
```

### 3. Insert an element at index 1.

```python
lists = [10, 20, 30]
print(lists)

lists.insert(1, 15)
print(lists)
```

### 4. Remove an element from a list.

```python
lists = [1, 2, 3, 4]
print(lists)

lists.remove(3)
print(lists)
```

### 5. Pop the last element from a list.

```python
lists = [1, 2, 3]
print(lists)

lists.pop(-1)
print(lists)
```

### 6. Find the length of a list.

```python
lists = [5, 10, 15]
print(lists)

print(len(lists))
```

### 7. Find the maximum element in a list.

```python
lists = [2, 8, 1, 9]
print(lists)

lists1 = max(lists)
print(lists1)
```

### 8. Find the minimum element in a list.

```python
lists = [5, 3, 7, 1]
print(lists)

lists1 = min(lists)
print(lists1)
```

### 9. Sort a list in ascending order.

```python
lists = [3, 1, 4, 2]
print(lists)

lists.sort()
print(lists)
```

### 10. Reverse a list.

```python
lists = [1, 2, 3]
print(lists)

lists1 = lists[ : : -1]
print(lists1)
```

### 11. Count how many times an element occurs.

```python
lists = [1, 2, 2, 3, 2]
print(lists)

lists1 = lists.count(2)
print(lists1)
```

### 12. Find the index of an element.

```python
lists = [10, 20, 30, 40]
print(lists)

lists1 = lists.index(30)
print(lists1)
```

### 13. Extend one list with another.

```python
lists1 = [1, 2]
print(lists1)

lists2 = [3, 4]
print(lists2)

lists1.extend(lists2)
print(lists1)
```

### 14. Copy a list.

```python
lists1 = [5, 10, 15]
print(lists1)

lists2 = lists1.copy()
print(lists2)
```

### 15. Clear all elements from a list.

```python
lists = [1, 2 ,3]
print(lists)

lists.clear()
print(lists)
```
