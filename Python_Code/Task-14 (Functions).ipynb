### 1. Write a function that takes a number as input and returns whether it is even or odd.

- Input: 12

- Output: "Even"

```python
def even_odd(num):
    if (num % 2 == 0):
        return "Even"
    else:
        return "Odd"
even_odd(12)
```

### 2. Write a function that accepts two numbers and returns the greater number.

```python
def maximum(num1, num2):
    if num1 > num2:
        return num1
    else:
        return num2

print("Greatest number is:", maximum(25, 40))
```

### 3. Write a function that takes a list of numbers and returns the sum of all numbers.

- Output: 20

```python
def sum1(numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total
numbers = [2,4,6,8]
print(numbers)
print(sum1(numbers))
```

```python
def product_num(products):
    result = 1
    for num in products:
        result = result * num
    return result
products = [1,3,5]
print(products)
print(product_num(products))
```

### 9. Write a function that accepts a number and returns the factorial of that number.

- Input: 5

- Output: 120

```python
def fact(number):
    factorial = 1
    for num in range (number, 0, -1):
        factorial = factorial * num
    return factorial

fact(5)
```

### 11. Write a function that accepts a number and returns the count of digits in it.

- Input: 98765

- Output: 5

```python
def count_digit(numbers):
    count = 0
    while numbers > 0:
        count = count + 1
        numbers = numbers//10
    return count

numbers = 98765
print(numbers)
print(count_digit(numbers))
```

### 12. Write a function that takes a string and returns whether it is a palindrome.

- Input: "madam"

- Output: True

```python
def palindrom(string):
    if string == string[::-1]:
        return True
    else:
        return False
            
print(palindrom("madam"))
```

### 13. Write a function that takes a number and returns whether it is a prime number.

- Input: 29

- Output: True

```python
def prime_num(num):
    if num <= 1:
        return False
    for i in range(2 ,num):
        if num % i == 0:
            return False
    return True
print(prime_num(29))
```

### 14. Write a function that takes two numbers and returns their greatest common divisor (GCD).

- Input: 36, 60

- Output: 12

```python
def gcd_number(num1, num2):
    result = 1
    for i in range(1, min(num1, num2) +1):
        if num1 % i == 0 and num2 % i == 0:
            result = i
    return result
print(gcd_number(36, 60))
```

### 15. Write a function that takes a list of numbers and returns a new list with only even numbers.

- Input: [1, 2, 3, 4, 5, 6]

- Output: [2, 4, 6]

```python
def list_num(numbers):
    result = []
    for i in numbers:
        if i % 2 == 0:
            result.append(i)
    return result
numbers = [1, 2, 3, 4, 5, 6]
print(numbers)
print(list_num(numbers))
```

### 16. Write a function that takes a list of numbers and returns a new list with only odd numbers.

- Input: [10, 11, 12, 13, 14]

- Output: [11, 13]

```python
def odd_num(numbers):
    result = []
    for i in numbers:
        if i % 2 != 0:
            result.append(i)
    return result
numbers = [10, 11, 12, 13, 14]
print(numbers)
print(odd_num(numbers))
```

### 17. Write a function that takes a list of numbers and returns the second largest number.

- Input: [15, 72, 9, 30, 50]

- Output: 50

```python
def large_num(numbers):
    first = second = float()
    for i in numbers:
        if i > first:
            second = first
            first = i
        elif i > second and i != first:
            second = i
    return second
numbers = [15, 72, 9, 30, 50]
print(numbers)
print(large_num(numbers))
```

### 18. Write a function that takes a string and returns the frequency of each character as a dictionary.

- Input: "hello"

- Output: {'h': 1, 'e': 1, 'l': 2, 'o': 1}

```python
def dict_frequency(string):
    result = {}
    for i in string:
        if i in result:
            result[i] += 1
        else:
            result[i] = 1
    return result
string = "hello"
print(string)
print(dict_frequency(string))
```

### 19. Write a function that accepts a number and returns a list of its multiplication table up to 10.

- Input: 5

- Output: [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

```python
def multiple_table(number):
    result = []
    for i in range(1, 10):
        result.append(number*i)
    return result 
number = 5
print(number)
print(multiple_table(number))
```

### 20. Write a function that takes two lists and returns a list of common elements.

- Input: [1, 2, 3, 4], [3, 4, 5, 6]

- Output: [3, 4]

```python
def common_ele(list1, list2):
    result = []
    for i in list1:
        if i in list2:
            result.append(i)
    return result
    
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
print(list1, list2)

print(common_ele(list1, list2))
```
