### 4. Given point = (10, 20, 30), change the value 20 to 200 by converting it to a list, modifying it, and converting it back to a tuple. Print the final tuple.

```python
point = (10,20,30)

# change the value 20 to 200
point1 = list(point)
print("Before Changing 20 to 200:", point1)

point1[1] = 200
print("After Changing 200:", point1)

tup = tuple(point1)
print("Converting to tuple:", tup)
```

### 13. Given a list dirty_data = ["A", "B", "A", "C", "B", "D"], convert it to a set to remove duplicates, then convert it back to a list and print it.

```python
dirty_data = ["A", "B", "A", "C", "B", "D"]
set1 = set(dirty_data)
print("Set:", set1)

dirty_data = list(set1)
print("List After Remove Duplicate Values:", dirty_data)
```

### 19. Given sentence = "apple banana apple cherry banana", split it into a list, convert it to a set to find unique words, and use a for loop to print each unique word.

```python
sentence = "apple banana apple cherry banana"
sent = sentence.split()
print(sent)

s = set(sent)
print(type(s))
for i in s:
    print(i)
```
