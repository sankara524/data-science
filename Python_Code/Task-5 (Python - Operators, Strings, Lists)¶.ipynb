### 1. Take two numbers and print:

```python
str1 = 10
str2 = 20
print(str1)
print(str2)

# Sum
sum1 = str1 + str2
print('Sum:',sum1) 

# Difference
diff = str1 - str2
print("Difference:", diff)

# Product
product = str1 * str2
print("Product:", product)

# Division
div = str1 / str2
print("Division:", div)
```
