### 1. Use map() with a function to get the length of each word in a list.
- Input: ['apple', 'banana', 'cherry']
- Output: [5, 6, 6]


```python
list1 =  ['apple', 'banana', 'cherry']

length = map(len, list1)
print(list(length))
```

### 2. Use filter() with a function to keep only strings longer than 3 characters.

Input: ['hi', 'hello', 'hey', 'world']
 
Output: ['hello', 'world']

```python
def string_longer(list1):
    if(len(list1) > 3):
        return list1

list1 = ['hi', 'hello', 'hey', 'world']
print(list1)

result = list(filter(string_longer, list1))
print(result)
```

### 3. Use reduce() with a function to find the maximum number in a list.
 
Input: [10, 25, 7, 99, 34]

Output: 99

```python
from functools import reduce

def max_num(i, j):
    return i if i>j else j
    
list1 = [10, 25, 7, 99, 34]
print(list1)

x = reduce(max_num, list1)
print(x)
```

### 4. Use filter() with a function to remove empty strings from a list.

Input: ['apple', '', 'banana', '', 'cherry']

Output: ['apple', 'banana', 'cherry']

```python
def remove_empty(list1):
    return len(list1) > 1

list1 = ['apple', '', 'banana', '', 'cherry']
print(list1)

result = list(filter(remove_empty, list1))
print(result)
```

### 5. Use map() with a function to convert temperatures from Celsius to Fahrenheit.

Input: [0, 25, 100]

Output: [32, 77, 212]

```python
def Celsius_to_Fahrenheit(c):
    return (c*9/5) + 32
    
temperature = [0, 25, 100]
print(temperature)

result = list(map(Celsius_to_Fahrenheit, temperature))
print(result)
```

### 7. Use a lambda with filter() to select numbers greater than 50.
 
Input: [10, 45, 60, 75, 30]

Output: [60, 75]

```python
def numbers(num):
    result = list(filter(lambda x : x > 50, num))
    return result
    
num = [10, 45, 60, 75, 30]
print(num)

print(numbers(num))
```

### 8. Use a lambda with map() to get the square of each number in a tuple.

Input: (3, 5, 7)

Output: [9, 25, 49]

```python
def sqr_num(tup):
    result = list(map(lambda x : x**2, tup))
    return result

tup = (3, 5, 7)
print(tup)

print(sqr_num(tup))
```

### 9. Use reduce() with a lambda to compute the product of all numbers.
 
 Input: [2, 3, 4, 5]
 
 Output: 120

```python
from functools import reduce

list1 = [2, 3, 4, 5]
print(list1)

result = reduce(lambda x, y : x*y, list1)
print(result)
```

### 10. Use map() with a lambda to title-case each string.
 
 Input: ['hello world', 'python programming']
 
 Output: ['Hello World', 'Python Programming']

```python
list1 = ['hello world', 'python programming']
print(list1)

result = list(map(lambda x : x.title(), list1))
print(result)
```

### 11. Use filter() with a lambda to keep palindromic strings.

Input: ['level', 'world', 'radar', 'hello', 'civic']

Output: ['level', 'radar', 'civic']

```python
list1 = ['level', 'world', 'radar', 'hello', 'civic']
print(list1)

result = list(filter(lambda x : True if (x == x[::-1]) else False, list1))
print(result)
```

### 12. Use reduce() with a lambda to concatenate a list of strings with a hyphen.

Input: ['a', 'b', 'c', 'd']
 
Output: 'a-b-c-d'

```python
from functools import reduce

list1 = ['a', 'b', 'c', 'd']
print(list1)

result = reduce(lambda x, y : x+"-"+y, list1)
print(result)
```

### 13. Use map() with a lambda to extract domains from emails (text after '@').
 
Input: ['alice@mail.com', 'bob@work.org', 'care@health.in']

Output: ['mail.com', 'work.org', 'health.in']

```python
list1 = ['alice@mail.com', 'bob@work.org', 'care@health.in']
print(list1)

result = list(map(lambda x : x[(x.index('@')+1) : ], list1))
print(result)
```

### 14. Use filter() with a lambda to keep numbers that are multiples of both 2 and 3.

Input: [3, 6, 9, 12, 15, 18, 20]

Output: [6, 12, 18]

```python
list1 = [3, 6, 9, 12, 15, 18, 20]
print(list1)

result = list(filter(lambda x :(x%2 == 0) and (x%3 == 0), list1))
print(result)
```

### 15. Use reduce() with a lambda to find the longest string (by length).
 
Input: ['ant', 'butterfly', 'cat', 'dinosaur']
 
Output: 'butterfly'

```python
from functools import reduce

list1 = ['ant', 'butterfly', 'cat', 'dinosaur']
print(list1)

result = reduce(lambda x, y : x if len(x) > len(y) else y, list1)
print(result)
```
