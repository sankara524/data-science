### 1. Read the file article.txt and print the entire content.

```python
with open("article.txt", "r")as file:
    print(file.read())
```

### 2. Count and print the total number of words in the article.

```python
with open("article.txt", "r") as file:
    data = file.read()
    words = data.split()
    print(len(words))
```

### 3. Count how many times each vowel appears in the article.

```python
vowels = "aeiou"
vowel_count = {}
for i in words:
    for j in i:
        if j in vowels:
            if j in vowel_count:
                vowel_count[j] += 1
            else:
                vowel_count[j] = 1
print(vowel_count)
```

### 4. Find the longest word in the article.

```python
longest_words = ""
for i in words:
    if len(i) > len(longest_words):
        longest_words = i
print(longest_words)
```

### 5. Print all sentences that contain a given keyword.

```python
sentence = data.split(".")
keyword1 = input("Enter: ")
for i in sentence:
    if keyword1 in i:
        print(i)
```

### 6. Count the number of words that start with a given letter.

```python
word = input()
count_words = 0
for i in words:
    if i.startswith(word):
        count_words += 1
print(count_words)
```

### 13. Convert the entire file content to uppercase and write it into upper.txt.

```python
with open("article.txt", "r") as file:
    with open("upper.txt", "a") as file2:
        for i in file:
            file2.write(i.upper())
```

```python
with open("upper.txt", "r") as file2:
    print(file2.read())
```
