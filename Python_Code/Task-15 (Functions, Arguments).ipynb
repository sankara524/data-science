### 1. Write a function that accepts a list of numbers and returns their sum.
 - Input: [1, 2, 3, 4, 5]
 - Output: 15

```python
def sum_num(numbers):
    result = 0
    for i in numbers:
        result += i
    return result
numbers = [1, 2, 3, 4, 5]
print(numbers)
print(sum_num(numbers))
```

### 2. Write a function that takes a string and returns it in reverse order.

```python
def reverse_str(string):
    return string[::-1]
string = "hello"
print(string)
print(reverse_str(string))
```

### 3. Write a function that takes a list of numbers and returns the maximum element.
 - Input: [3, 8, 1, 6, 2]
 - Output: 8

```python
def maximum(num):
    maxi = num[0]
    for i in num:
        if i > maxi:
            return i
    return maxi
num = [3, 8, 1, 6, 2]
print(num)
print(f"Maximum Element: {maximum(num)}")
```

### 4. Write a function that accepts a number and returns whether it is prime or not.
 
 Input: 7
 
 Output: True

```python
def prime_num(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

num = 7
print(num)
print(prime_num(num))
```

### 5. Write a function that accepts a string and returns the number of vowels in it.
 
 Input: "programming"
 
 Output: 3

```python
def count_vowel(string):
    count = 0
    vowels = 'aeiou'
    for i in string:
        if i in vowels:
            count = count + 1
    return count

string = "programming"
print(string)

print(count_vowel(string))
```

### Q6. Write a function that accepts two numbers and returns their greatest common divisor (GCD).

 Input: a = 12, b = 18

 Output: 6

```python
def great(a, b):
    gcd = 1
    for i in range(1, min(a,b) +1):
        if (a%i == 0) and (b%i == 0):
            gcd = i
    return gcd

a = 12
b = 18
print(a, b)

print(great(a, b))
```

### 8. Write a function that accepts a dictionary and returns the key with the maximum value.
 
 Input: {'a': 10, 'b': 20, 'c': 5}
 
 Output: 'b'

```python
def maximum_value(data):
    max_key = None
    max_value = float('-inf')

    for key, value in data.items():
        if value > max_value:
            max_value = value
            max_key = key

    return max_key

data = {'a': 10, 'b': 20, 'c': 5}
print(data)

print(maximum_value(data))
```

### 9. Write a function that accepts a number n and returns the factorial of n.
 
 Input: 5
 
 Output: 120

```python
def factorial_num(n):
    if n < 0:
        return "This is negative number"
    factorial = 1
    for i in range(1, n+1):
        factorial *= i
    return factorial

n = 5
print(n)

print(factorial_num(n))
```

### 10. Write a function that accepts a string and returns a dictionary with the frequency of each character.
 
 Input: "apple"
 
 Output: {'a': 1, 'p': 2, 'l': 1, 'e': 1}

```python
def dict_frequency(data):
    frequency = {}
    for char in data:
        frequency[char] = frequency.get(char, 0) + 1
    return frequency

data = "apple"
print(data)

print(dict_frequency(data))
```

### 17. Write a function multiply(*numbers) that returns the product of all numbers passed to it. Call it with 3 and 5 arguments.
 
 Input: multiply(2, 3, 4)
 
 Output: 24
 
 Input: multiply(1, 2, 3, 4, 5)
 
 Output: 120

```python
def multiply(*numbers):
    product = 1
    for n in numbers:
        product *= n
    return product

print(multiply(2, 3, 4))
print(multiply(1, 2, 3, 4, 5))
```

### 18. Write a function print_kwargs(**kwargs) that prints all key-value pairs passed as keyword arguments. Call it with 1 and 3 key-value pairs.
 
 Input: print_kwargs(name="Alice")
 
 Output: name: Alice
 
 Input: print_kwargs(name="Bob", age=25, city="Delhi")
 
 Output:

name: Bob

age: 25

city: Delhi

```python
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_kwargs(name="Alice")
print()
print_kwargs(name="Bob", age=25, city="Delhi")
```

### 19. Write a function sum_numbers(*numbers, start=0) that returns the sum of all numbers plus start. Call it with numbers only and with start value.
 
 Input: sum_numbers(1, 2, 3)
 
 Output: 6
 
 Input: sum_numbers(1, 2, 3, start=5)
 
 Output: 11

```python
def sum_numbers(*numbers, start=0):
    return sum(numbers) + start

print(sum_numbers(1, 2, 3))
print(sum_numbers(1, 2, 3, start=5))
```

### 20. Write a function list_to_string(*items, sep="-") that joins all items using the separator. Call it with default and custom separator.
 
 Input: list_to_string("a", "b", "c")
 
 Output: "a-b-c"

 Input: list_to_string("1", "2", "3", sep=":")

 Output: "1:2:3"

```python
def list_to_string(*items, sep="-"):
    return sep.join(items)

print(list_to_string("a", "b", "c"))
print(list_to_string("1", "2", "3", sep=":"))
```
