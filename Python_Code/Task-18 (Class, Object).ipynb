### 1 : Book Class

- Create a class Book with attributes: title, author, price
    
- Add a constructor to initialize these attributes
 
- Add a method display_info() to print book details

- Create 2 book objects and display their information


```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_info(self):
        print(f"Title: {self.title}, Author: {self.author}, Price: {self.price}")

book1 = Book("Python", "Guido van Rossum", 500)
book1.display_info()

book2 = Book("Java", "James Gosling", 500)
book2.display_info()
```

### 2 : Employee Salary Calculator

- Create a class Employee with attributes: name, salary, department

- Add a method annual_salary() that returns salary × 12

- Add a method give_raise(percentage) that increases salary by given percentage

- Create 2 employees, display their annual salary, give them a 10% raise, and display new annual salary

```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def annual_salary(self):
        return self.salary*12

    def give_raise(self, percentage):
        self.salary = self.salary + (self.salary * percentage / 100)


employee1 = Employee("Kiran", 100000, "DS") 
employee2 = Employee("Ramu", 80000, "DS")

print(employee1.name,": Before annual salary Raise:", employee1.annual_salary())
print(employee2.name,": Before annual salary Raise:", employee2.annual_salary())
print("\n")
      
employee1.give_raise(10)
employee2.give_raise(10)

print(employee1.name,": After annual salary Raise:", employee1.annual_salary())
print(employee1.name,": After annual salary Raise:", employee2.annual_salary())
```

### 3 : Rectangle Class

- Create a class Rectangle with attributes: length, width

- Add methods:
    
  area() → returns length × width

  
  perimeter() → returns 2 × (length + width)

  
  is_square() → returns True if length equals width, else False

- Create 3 rectangles and display their area, perimeter, and whether they're squares

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2*(self.length * self.width)

    def is_square(self):
        if (self.length == self.width):
            return True
        else:  
            return False

rectangle1 = Rectangle(10, 20)
rectangle2 = Rectangle(15, 30)
rectangle3 = Rectangle(6, 6)

print("Area of Rectangle:", rectangle1.area())
print("Perimeter of Rectangle:", rectangle2.perimeter())
print("Is square :", rectangle3.is_square())
```

### 4 : Temperature Converter

- Create a class Temperature with attribute: celsius

- Add methods:
    
    to_fahrenheit() → returns (celsius × 9/5) + 32
    
    to_kelvin() → returns celsius + 273.15
    
    is_freezing() → returns True if celsius ≤ 0

- Create 3 temperature objects and display conversions

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9/5) + 32

    def to_kelvin(self):
        return (self.celsius + 273.15)

    def is_freezing(self):
        if (self.celsius <= 0):
            return True
        else:
            return False

temperature1 = Temperature(10)
temperature2 = Temperature(10)
temperature3 = Temperature(10)

print("Temperature to Fahrenheit:", temperature1.to_fahrenheit())
print("Temperature to kelvin:", temperature2.to_kelvin())
print("Temperature to Freezing:", temperature3.is_freezing())
```

### 5 : Movie Rating System

- Create a class Movie with attributes: title, director, rating (out of 10)

- Add methods:
    
    display_info() → prints movie details
    
    rating_category() → returns "Excellent" if rating ≥ 8, "Good" if ≥ 6, "Average" if ≥ 4, "Poor" if < 4
                                                                                          
    is_blockbuster() → returns True if rating ≥ 8

- Create 4 movie objects with different ratings and display their category and blockbuster status

```python
class Movie:
    def __init__(self, title, director, rating):
        self.title = title
        self.director = director
        self.rating = rating

    def display_info(self):
        print(f"Movie Title: {self.title}, Director: {self.director}, Rating(Out of 10): {self.rating}")

    def rating_category(self):
        if(self.rating >= 8):
            return "Excellent"
        elif(self.rating >= 6):
            return "Good"
        elif(self.rating >= 4):
            return "Average"
        else:
            return "Poor"
     
    def is_blockbuster(self):
        if(self.rating >= 8):
            return True
        else:
            return False

movie1 = Movie("A", "B", 10)
movie2 = Movie("B", "C", 7)
movie3 = Movie("C", "D", 4)
movie4 = Movie("D", "E", 3)

movies = [movie1, movie2, movie3, movie4]
for m in movies:
    m.display_info()
    print("Movie Rating Category:", m.rating_category())
    print("Is Movie Blockbuster:", m.is_blockbuster())
    print("\n")
```
