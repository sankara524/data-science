### Type Conversion
> Converting from one data type to another

> type_fun(variable)

```python
int_num = 10
print(type(int_num))
float_num = float(int_num)
print(float_num, type(float_num))
```

```python
# Change type from float to string
print(float_num)
str_num = str(float_num)
print(str_num, type(str_num))
```

```python
# Change type from string to float
print(str_num)
float_num = float(str_num)
print(float_num, type(float_num))

str_name ='Kiran'
float_num = float(str_name)
print(float_num, type(float_num))
```

**Note:** For type conversion from String to float/int, variable should be any int/float with data type String.

```python
# Change type from float to boolean
print(float_num)
bool_num = bool(float_num)
print(bool_num, type(bool_num))

float_num = 0.0
print(float_num)
bool_num = bool(float_num)
print(bool_num, type(bool_num)) 
```

**Note:** Any Number Greater than 0 would be True else False. 

```python
# Change type from boolean to string
print(bool_num)
str_name = str(bool_num)
print(str_name, type(str_name))
```

```python
# Change type from string to boolean
print(str_name)
bool_name = bool(str_name)
print(bool_name, type(bool_name))

str_name = ''
print(str_name)
bool_name = bool(str_name)
print(bool_name, type(bool_name))
```

**Note:** Any String other than blank space would be True or False.

## **Opearators and Expression**

#### Operator:

- c = a + b (a,b : Variables)
- d = 3 + 8 (3,8 : literals/values)
- e = c
- symbols : = , +
> Any symbol that performs actions/operatins is called ***Operator***.

### Operand:

>Variables or literals used in an operation is called ***Operand***.
- **e.g.** a,b,3,8

## **Expression**:
> Multiple operands and operator togethr performimg operation becomes ***Expression***.
- **e.g.**  a+b-3+8

### Arithmatic Operators:
, -, *, /, // (floor Division), % (Modulus), ** (exponent)

```python
# Addition
a = 6
b = 5
print('a+b is:', a+b)

# Subtraction
print('a-b is:', a-b)

# Multiplication
print('a*b is:', a*b)
      
# Division
print('a/b is:',a/b)
```

```python
# Floor Divsion
print(a, b)
print('a/b:', a/b)   # Normal division
print('a//b:', a//b)  # Floor Division
```

```python
# Remainder 
print(a,b)
print('Remaider of a%b is:', a%b)
```

```python
# Exponent
print(a, b)
print('a**b is:', a**b)
```

### Comparison Operator : Relational Operator
<, >, == (equality operator), !=, >=, <=

Output: True/False (Boolean Value) 

```python
# Less than operator
print('a < b is:', a<b)

# Greater than operator
print('a > b is:', a>b)

# Equality operator
print('a == b is:', a==b)

# Not Equality operator


# Greater than Equal to operator
print('a >= b is:', a>=b)

# Less than Equal to operator
print('a <= b is:', a<=b)
```

### Logical Operator
> and, or, not

Output: True/False (Boolean value)

```python
x = True
y = False

# and opetor 
print('x and y is:', x and y)

# or operator
print('x or y is:', x or y)

# Not operator
print('not x is:', not x)
```

```python
x = 2
y = 3

# and operator 
print('x == 2 and y == 3 is:', x == 2 and y == 3)

# or operator
print('x == 2 or y == 4 is:', x == 2 or y == 4)

# Not operator
print('not x is:', not x)
```
