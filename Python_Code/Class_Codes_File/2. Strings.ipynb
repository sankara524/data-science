### String Formating

- Very important
  
**14.1.  .format()** : Used us a dynamic Entry in a string.

**14.2. f String literals**

```python
# Static String

print('Today is Tuesday and it is cloudy')
```

```python
# Dynamic String
# Passing the values to the placeholders "{}" using the default order.

print('Today is {} and it is {}'.format('Tuesday', 'cloudy'))  # {} : place holders.
```

```python
# Grammatical Error with default order.

print('Today is {} and it is {}'.format('cloudy', 'Tueday'))
```

```python
# Passing the value to the palceholders "{}" using the positional order'

print('Today is {1} and it is {0}'.format('cloudy', 'Tueday'))
```

#### f String literals

```python
print(f"Today is {'tuesday'} and it is {'cloudy'}")
```

```python
print(f"Today is {'wednesday'} and it is {'rainy'}")
```

```python
day = 'tuesday'
condition = 'rainy'

print(f"Today is {day} and it is {condition}")
```
