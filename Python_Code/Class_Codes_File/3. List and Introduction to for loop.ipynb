## Built-in Methods or function

### 1. Append()

- **append():** is used to add a single element to the end of the of the list.

```python
# define a list 1 to 9
num_lst = list(range(1,10))

# add the value, 10, to the end of the list

# print with a note
print('num_lst before:', num_lst)

# Call method
num_lst.append(10)

# print with a note
print('num_lst After adding 10:', num_lst)

# Check length of list 
print(len(num_lst))
```

### 2. extend()
- **extend()** : is used to extend our current list with another list

```python
# Define list 11 to 15
num_lst2 = list(range(11,16))
print(num_lst2)

# Print num_lst before extending
print('num_lst Before extending:', num_lst)

# Call method
num_lst.extend(num_lst2)  # num_list = primary and num_lst2 = Secondary

# Print after extending
print('num_lst After extending:', num_lst)

# check length
print(len(num_lst2))
```

```python
# Define list 11 to 15
num_lst2 = list(range(11,16))
print(num_lst2)

# Print num_lst before extending
print('num_lst Before extending:', num_lst)

# Call method
num_lst = num_lst.extend(num_lst2)  # num_list = primary and num_lst2 = Secondary

# Print after extending
print('num_lst After extending:', num_lst)

# check length
print(len(num_lst2))
```

#### **Note** : One list can be extended with the other by cancatenating.

```python
# Another way to add the elements: by cancatenating
print(num_lst)
print(num_lst + num_lst2)
```

### Shallow copy: copy()

- Creates a new object, but the contents(nested objects) are still reference to the original.

```python
# To create one more instance of a list
# Define list 1
l1 = [1,2,[3,4]] # nested list
print(f"l1: {l1}")

# Call library and function
l2 = copy.copy(l1)
print(f"l2: {l2}")

# Assign value to second list

# l2[2]   # normal index
l2[2][0] = 33  # nested index

#print both
print(l1)
print(l2)
```

### deepcopy()
- Create a new object, and recursively copies all nested objects.

```python
# Define list 1
l1 = [3,4,5,6,7]
print("l1:", l1)

# Call library and function
l2 = copy.deepcopy(l1)
print("l2:", l2)

# Assign value to second list
l2[1] = 10

# print both
print(l1)
print(l2)
```

```python
# nested list
l1 = [3,4,[5,8],6,7]
print("l1:", l1)

# call library and function
l2 = copy.deepcopy(l1)
print("l2:", l2)

# Assign value to second list
l2[2][0] = 10

# print both
print(l1)
print(l2)
```

### **Rules** :

- Use **shallow copy** if your data structure is flat (no nested mutable objects).
- Use **Deep copy** if you have nested/multi-level structures (lists inside a lists, dicts inside a dicts, etc.).

### **Simple for loop Example**

Lets loop through a list of fruits and print each one.

We are using:

- A list of fruits
- A **for loop** to iterate(repeat) through each fruit
- A print statement to display a message

This is a beginner-friendly example to understand the **for loop** in Python.

```python
# Define list fruits
fruits = ["Apple", "Orange", "Kiwi"]
print(f'fruits:{fruits}')

# Run loop
for x in fruits:
    print(x)

# Run loop with a message
for x in fruits:
    print(f'I like fruits such as: {x}')
```

### **Undertanding range() in python**

The **range()** function generates a sequence of numbers.

- Used commonly in loops.
- Syntax: **range(start, stop, step)**
- If only one argument is given, it's treated as the stop.

```python
# Print 1 to 10 for using loop
for i in range(1, 11):
    print(i)
```

```python
# Print 1 to 10 for using loop
for i in range(1, 11):
    print(i, end = '\n')  # "\n" : new line (it is a defualt)

for i in range(1, 11):
    print(i, end = ' ')  # print side by side.  
```

```python
# range without loop
list(range(1,11))
```

**Note:** Useful for controlled, repeating tasks, or indexing.

enumerate() helps access both index and value during iteration.
