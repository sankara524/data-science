# Tuple
- Collection of items which can be indexed but immutable, (apple, 250)
- So once we declare the tuple, we can not edit the data
- tuple will be declared using `()`

### tuple declaration

```python
tuple_num = (10, 5, 2)
print(tuple_num)
print(type(tuple_num))
print(len(tuple_num))
```

```python
tup = ()
print(tup)
print(type(tup))
print(len(tup))
```

### How to add elements in tuple?

```python
tuple1 = (1,2,3)
tuple2 = (4,5,6)

# to Add
tuple3 = (tuple1 + tuple2)
tuple3
```

### Built-in Funtions

- In tuple we have only 2 built in funtions because the tuple is Immutable.

- **1. count():** Number of Occurence of an item in a tuple.
- **2. index():** Positional value on an item in a tuple.

```python
tup = (3,2,6,3)

# count of 3
print(f'Number of 3 in tup: {tup.count(3)}')

# index of 1st 3
tup.index(3)

# index of last 3
tup.index(-1) 
```

```python
tup = (3,2,6,3)

# count of 3
print(f'Number of 3 in tup: {tup.count(3)}')

# index of 1st 3
idx3 = tup.index(3)

# index of last 3
print("Hardcoded:", tup.index(3,1))  # static / Hardcoded

# index of last 3
print("Softcoded:", tup.index(3, idx3+1))  # Dynamic / Softcoded
```

### Indexing and Slicing

```python
tup1 = (1,2,3,4,5,6)

# fetch 6
tup1[-1]
```

```python
# fetch from 3 to 5

tup1 = (1,2,3,4,5,6)

tup1[2 : 5]  # Hardcoded
tup1[tup1.index(3) : tup1.index(5)+1] # Softcoded
```

```python
tup1 = (1,2,3,4,5,6)

# Change 4 to 44
print("Before 44:", tup1)

# Convert to a list
list1 = list(tup1)
print("After converting to list:", list1)

# Change 4 to 44
list1[list1.index(4)] = 44
print("After Changing 4 to 44:", list1)

# convert list to tuple
tup1 = tuple(list1)
print("After 44:", tup1)
```

### packing and unpacking

```python
# Input
tup1 = ('a', 'b', 'c')
tup2 = (1,2,3)
zip(tup1, tup2)
```

```python
# Output with list
tup3 = tuple(list(zip(tup1, tup2)))
tup3
```

```python
# ouput without list it also working
tup3 = tuple(zip(tup1, tup2))
tup3
```

```python
# Tuple packing
tup1 = (10,)
tup2 = (20,)
tup3 = tup1 + tup2
tup3
```

```python
# Tuple Unpacking
a, b = tup3
print(a)
print(b)
```

# Sets

- **Sets are like lists in that you use them for storing a collection of elements. Unlike lists, however, the elements in a set are nonduplicates and are not placed in any particular order.**
- **with { } we declare a set**
- **sets are unordered and unindexed**
- **set will not allow duplicate elements**

```python
# Stes Declaration
set1 = {1,2,3,3}
print(f'set is: {set1}')
print(type(set1))
print("len of set:", len(set1))  # No Duplicate values, so len is 3. The 2. 3's are removed by 1.
```

```python
set2 = {1,2,3,3, 'hi', True, 'hi'}  # True == 1, False == 0
print(f'set is: {set2}')
print(type(set2))
print("len of set:", len(set2))
```

```python
set3 = {1,2,3,3, 'hi', True, 'hi', False, 2.5, 2.5}   # True == 1
print(f'set is: {set3}')   # True Acts as 1. That's why True has not print. 
print(type(set3))
print("len of set:", len(set3))
```

```python
set3 = {0,1,2,3,3, 'hi', True, 'hi', False, 2.5, 2.5}   # False == 0
print(f'set is: {set3}')     # False acts zero. Thats why False has not print.
print(type(set3))
print("len of set:", len(set3))
```

```python
set3 = {2,3,3, 'hi', True, 'hi', False, 2.5, 2.5}
print(f'set is: {set3}')
print(type(set3))
print("len of set:", len(set3))
```

```python
set3 = {True,1,2,3,3, 'hi', True, 'hi', False, 2.5, 2.5}
print(f'set is: {set3}')
print(type(set3))
print("len of set:", len(set3))
```

# Other set operations
- 1. union()
- 2. intersection()
- 3. difference()
-  4. Symmetrical Difference()

### 1. union()

```python
#Sets union, intersection and difference
A = {2,3,4}
B = {4,5,6}
print("A union B:", A|B)
print("B union A:", B|A)

print("Alternative Method of A Union B:", A.union(B))
```

### 2. intersection()

```python
#Sets union, intersection and difference
A = {2,3,4}
B = {4,5,6}

print("A intersection B:", A & B)
print("B intersection A:", B & A)

print("Alternative Method of A intersection B:", A.intersection(B))
```

```python
a ,b = {1,2,3}, {4,5,6}
a & b
```

### 3. difference()

```python
A = {2,3,4}
B = {4,5,6}

print("A - B", A-B)
print("B - A", B-A)
print("Alternative Method of A Difference B:", A.difference(B))
print("Alternative Method of B Difference A:", B.difference(A))
```

### 4. symmetric difference

```python
A = {2,3,4}
B = {4,5,6}

print(A)
print(B)

#Symmetric difference prints values of A and B and eliminates the common
#elements from both the sets

#This can also be written with union and intersection
# Union
print("A Union B:", A.union(B))

# intersection
print("A Intersection B:",A.intersection(B))

# Symmetric Difference
print("A Symmetric Difference B:", A ^ B)
print("A Alternative Symmetric Difference B:", A.symmetric_difference(B))
```

# 2. Manipulating and Accessing Sets
- You can add an element to a set or remove an element by using the add(e) or remove(e) method. You can use the len, min, max, and sum functions on a set, and a for loop to traverse all elements in a set. You can use the in or not in operator to determine whether an element is in the set.

```python
# Length of set
set2 = {2,4,5,7,8}
print(f"length of set: {len(set2)}")  # Use function len

# Minimum of set
print(f'Minimum of set: {min(set2)}')  # Use min

# Maximum of set
print(f'Maximum of set: {max(set2)}')  # Use max

# Sum of set
print(f"Sum of set: {sum(set2)}")  # Use sum
```
