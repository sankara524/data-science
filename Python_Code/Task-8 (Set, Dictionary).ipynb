## **Task-8 (Set, Dictionary)**

## **Part A: Set**


### 1. Find common elements from three sets.

```python
set1 = {1,2,3,4}
set2 = {2,3,5,6}
set3 = {0,2,3,7}

set4 = set1.intersection(set2, set3)
print(set4)
```

### 2. Find elements that are only in the first set but not in the other two.

```python
set1 = {1,2,3,4,5}
set2 = {2,3}
set3 = {4,6}

set4 = set1.difference(set2,set3)
print(set4)
```

### 3. Create a set of unique vowels from a string.

```python
str1 = "education"
str2 = (set(str1) & {'a', 'e', 'i', 'o', 'u'})
print(str2)
```

### 4. Convert a list of tuples into a set.

```python
lists1 = [(1, 2), (3, 4), (1, 2)]
lists2 = set(lists1)
print(lists2)
```

### 5. Find the intersection of characters in two words.

```python
set1 = set("banana")
set2 = set("bandana")
set3 = set1.intersection(set2)
print(set3)
```

### 6. Find digits in the first number but not in the second.


```python
set1 = set("12345")
set2 = set("34567")

set3 = set1 - set2
print(set3)
```

### 7. Find all unique words from a sentence.

```python
str1 = "python is fun and python is powerful"
word = str1.split()
str2 = set(word)
print(str2)
```

### 8. Remove duplicates from a list.

```python
lists1 = [1,2,2,3,3,4,5]
lists2 = set(lists1)
print(lists2)
```

### 9. Find non-overlapping characters between two words.

```python
set1 = set("hello")
set2 = set("world")

set3 = set1.symmetric_difference(set2)
print(set3)
```

### 10. Find maximum and minimum elements in a set.

```python
set1 = {12, 7, 25, 3, 18}

# Maximum
set2 = max(set1)
print("Maximum Number:", set2)

# Minimum
set3 = min(set1)
print("Minimum Number:", set3)
```

## **Part B: Dictionary**

### 1. Get the value of a dictionary with the maximum value.

```python
data = {
    'a' : 10,
    'b' : 25,
    'c' : 15 
}

data1 = max(data.values())
print(data1)
```

### 2. Get all dictionary values as a single string.

```python
data = {
    'a' : 'hi',
    'b' : 'there'
}
data1 = " ".join(data.values())
print(data1)
```

### 3. Remove the last inserted pair.

```python
data = {'a' : 1, 
        'b' : 2, 
        'c' : 3
}
data.popitem()
print(data)
```

### 4. Find values present in both dictionaries.

```python
data1 = {'x':1, 
        'y':2}
data2 = {'a':2,
         'b':3}

result = set(data1.values()) & set(data2.values())
print(result)
```

### 5. Find the sum of dictionary keys.

```python
data = {10:'a',
        20:'b', 
        30:'c'}
result = sum(data.keys())
print(result)
```

### 6. Remove duplicates from dictionary values.

```python
data = {'a':1,
        'b':2,
        'c':1}
result = set(data.values())
print(result)
```
