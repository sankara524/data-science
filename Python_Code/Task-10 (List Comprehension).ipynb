# **Task-10 (List Comprehension)**

### 1. If a list comprehension generates the squares of numbers from 1 to 4, is the result [1, 4, 9, 16]?

```python
sqr_num = list(range(1, 5))
print(sqr_num)

result = [i ** 2 for i in sqr_num]
print(result)
```

### 2. Using a list comprehension to collect all even numbers between 0 and 9, is the length of the resulting list 5?

```python
length = list(range(0, 10))
print(length)

result = [i for i in length if i % 2 == 0]
print("Length:", len(result))
print(result)
```

### 3. If you add 5 to each element in the list [1, 3, 5] using a list comprehension, is the resulting list [6, 8, 10]?

```python
list1 = [1,3,5]
print(list1)

result = [i + 5 for i in list1 ]
print(result)
```

### 4. A list comprehension multiplies each odd number from 1 to 9 by 3. Is the resulting list [3, 9, 15, 21, 27]?

```python
odd_num = list(range(1, 10))
print(odd_num)

result = [i*3 for i in odd_num if i%2 != 0]
print(result)
```

### 5. If you use a list comprehension to filter values greater than 6 from the list [2, 6, 7, 9, 10], is the sum of the resulting list 26?

```python
list2 = [2,6,7,9,10]
print(list2)

result = [i for i in list2 if i>6]
print(result)
print(sum(result))
```

### 6. If a list comprehension generates all numbers divisible by 3 from 1 to 15, is the result [3, 6, 9, 12, 15]?

```python
list3 = list(range(1 , 16))
print(list3)

result = [i for i in list3 if i % 3 == 0]
print(result)
```

### 7. Using a list comprehension to extract the first letter of each word in ["apple", "banana", "cherry"], is the resulting list ['a', 'b', 'c']?

```python
word = ["apple", "banana", "cherry"]
print(word)

letter = [i[0] for i in word]
print(letter)
```

### 8. If a list comprehension creates [x*2 for x in range(4)], is the result [0, 2, 4, 6]?

```python
list4 = [0,1,2,3]
print(list4)

result = [x*2 for x in range(4)]
print(result)
```

### 9. Using a list comprehension to convert ["a", "b", "c"] into uppercase, is the resulting list ['A', 'B', 'C']?

```python
list5 = ["a", "b", "c"]
print(list5)

result = [i.upper() for i in list5]
print(result)
```

### 10. If a list comprehension filters out numbers less than 5 from [1, 4, 6, 8, 3, 10], is the resulting list [6, 8, 10]?

```python
list6 = [1, 4, 6, 8, 3, 10]
print(list6)

result = [i for i in list6 if 5 < i]
print(result)
```
