### 1. Given fruits = ["apple", "banana", "cherry"], use a for loop to print each fruit in uppercase.

```python
# To Upper Case
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x.upper())
```

### 10. Given cities = ["New York", "Delhi", "Mumbai", "London"], use a for loop and if to print only the cities that start with 'D'.

### 11. Given data = "apple,banana,orange", first .split(",") it into a list, then use a for loop to print:
 **Fruit: apple (using an f-string).**


```python
data = "apple,banana,orange"
fruits = data.split(",")
print(fruits)

for d in fruits:
    print(f'Fruit: {d}')
```

### 12. Given sentence = "I love coding in python", split it into words and use a for loop and if to print only words with more than 4 characters.

```python
sentence = "I love coding in python"
sent = sentence.split(" ")
print(sent)

for s in sent:
    if(len(s) > 4):
        print(s)
```

### 13. Given binary_list = [1, 0, 1, 1, 0], use a for loop and if to print "Found Active" every time the value is 1.

```python
binary_list = [1, 0, 1, 1, 0]
for binary in binary_list:
    if binary == 1:
        print("Found Active:", binary)
    else:
        print("Not Found:", binary)
```

### 15. Given numbers = [1, 2, 3, 4, 5], use a for loop to calculate the total sum and print it at the end using an f-string:
 **The final total is: 15**

```python
numbers = [1, 2, 3, 4, 5]
total_sum = 0

for num in numbers:
    total_sum = total_sum + num
    print(total_sum)
print(f'The final total is : {total_sum}')
```
