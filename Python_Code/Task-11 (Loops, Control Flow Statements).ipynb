### 1. Print numbers from 1 to 20 but skip multiples of 3 and 5 (use while + transfer statements).

```python
i = 0
while(i <= 20):
    print(i)
    if(i == 3 and i == 5):
        print(i)
        break
```

### 2. Write a program that checks all characters in a string, but print only consonants. Print only consonants (use for + transfer statements).
- Input: "python"

- Output: p y t h n

```python
string1 = "python"
print(string1)

vowels = "aeiou"
for i in string1:
    if i in vowels:
        continue
    print(i, end=" ")
```

### 3. Keep multiplying numbers from a list until the product exceeds 500, then break (use for + transfer statements).
- Input: [2, 5, 10, 3, 6]

```python
num = [2, 5, 10, 3, 6]
product = 1

for i in num:
    product = product*i
    print(product)
    if (product >= 500):
        print("exceeds 500 stop")
        break
    
```

### 4. Traverse through a list of integers. Skip negative numbers, but stop when you find a 0 (use for + transfer statements).
- Input: [10, -3, 5, -7, 0, 8]
- Output: 10 5

```python
number = [10, -3, 5, -7, 0, 8]
print(number)

for i in  number:
    if (i < 0):
        continue    
    elif (i == 0):
        break    
    print(i, end=" ")    
```

### 5. Find the sum of numbers from 1 to 50 using a while loop.
- Output: 1275

```python
i = 1
total = 0 

while i <= 50:
    total = total + i
    i = i+1
print(total)
```

### 7. Reverse a number using a for loop (don’t use str()).
- Input: 1234 
- Output: 4321

```python
num = 1234
print(num)

reverse = 0
for i in range(10):
    if num == 0:
        break
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print(reverse)
```

### 8. Find the sum of digits of a number using for loop (don’t use str()).
- Input: 1234 
- Output: 10


```python
number = [1,2,3,4]
print(number)
sum1 = 0
for i in number:
    sum1 = sum1 + i
    print(sum1)
print("Sum of digits of a number:", sum1)  
```
