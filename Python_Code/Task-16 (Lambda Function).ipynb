## **Task-16 (Lambda Function)**

### 1. Write a lambda function to add two numbers.
 - Input: 5, 7
 - Output: 12


```python
num = lambda a, b : a + b
print(num(5,7))
```

### 2. Write a lambda function to find the square of a number.
 - Input: 6
 - Output: 36


```python
sqr_num = lambda a : a ** 2
print(sqr_num(6))
```

### 3. Write a lambda function to find the maximum of two numbers.
 - Input: 10, 25
 - Output: 25

```python
maximum = lambda a, b : max(a, b)
print(maximum(10, 25))
```

### 4. Write a lambda function that returns True if a number is even, otherwise False.
 - Input: 9
 - Output: False

```python
even_odd = lambda i : True if i % 2 == 0 else False
print(even_odd(9))
```

### 5. Write a lambda function to get the last character of a string.
 - Input: "python"
 - Output: "n"

```python
str1 = "python"
print(str1)
last = lambda x : x[-1]
print(last(str1))
```

### 6. Use a lambda function with map() to double each element in the list.
 - Input: [1, 2, 3, 4]
 - Output: [2, 4, 6, 8]

```python
list1 = [1, 2, 3, 4]
print(list1)

elements = list(map(lambda x : x * 2, list1))
print(elements)
```

### 7. Use a lambda function with filter() to extract only odd numbers from a list.
 - Input: [1, 2, 3, 4, 5, 6]
 - Output: [1, 3, 5]

```python
list1 = [1, 2, 3, 4, 5, 6]
print(list1)

elements = list(filter(lambda x : x % 2 != 0, list1))
print(elements)
```

### 8. Use a lambda function with map() to convert a list of strings to uppercase.
 - Input: ["apple", "banana", "kiwi"]
 - Output: ["APPLE", "BANANA", "KIWI"]

```python
list1 = ["apple", "banana", "kiwi"]
print(list1)

upper_case = list(map(lambda x : x.upper(), list1))
print(upper_case)
```

### 9. Use a lambda function with filter() to get words that start with the letter 'a'.
 - Input: ["apple", "banana", "avocado", "grape"]
 - Output: ["apple", "avocado"]

```python
list1 = ["apple", "banana", "avocado", "grape"]
print(list1)

words = list(filter(lambda x : x.startswith("a"), list1))
print(words)           
```

### 10. Use reduce() with a lambda function to find the product of all numbers in a list.
 - Input: [1, 2, 3, 4]
 - Output: 24

```python
from functools import reduce

list1 = [1, 2, 3, 4]
print(list1)

number = reduce(lambda a, b : a * b, list1)
print(number)
```
